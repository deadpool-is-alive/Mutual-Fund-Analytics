```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path("../data/raw")

csv_files = list(DATA_DIR.glob("*.csv"))

print(f"Found {len(csv_files)} CSV Files")
```

```python
def inspect_dataset(file_path):
    df = pd.read_csv(file_path)

    print("=" * 80)
    print(f"Dataset: {file_path.name}")
    print("=" * 80)

    print(f"Shape: {df.shape}")

    print("\nColumns: ")
    print(df.columns.tolist())

    print("\nData Types: ")
    print(df.dtypes)

    print("\nMissing Values: ")
    print(df.isnull().sum())

    print("\nDuplicate Rows: ")
    print(df.duplicated().sum())

    print("\nHead: ")
    display(df.head())

    return df
```

```python
datasets = {}

for file in csv_files:
    datasets[file.stem] = inspect_dataset(file)
```

```python
df_nav_history = pd.read_csv(DATA_DIR / "02_nav_history.csv")
```

```python
df_nav_history.sample(5)
```

```python
def validating_codes(df_left, df_right):
    validation_df = pd.merge(df_left, df_right, on='amfi_code', how='left', indicator=True)

    invalid_records = validation_df[validation_df['_merge'] == 'left_only']

    if not invalid_records.empty:
        print(f"Warning: Found {len(invalid_records)} invalid amfi_code(s) in the left DataFrame.")
        print(invalid_records[['amfi_code', 'fund_name', '_merge']])
        return
    
    print("All amfi_code(s) in the left DataFrame are valid and present in the right DataFrame.")
```

```python
validating_codes(df_nav_history, df_fund_master)
```
